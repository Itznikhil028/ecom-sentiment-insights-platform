```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot appearance background clear karne ke liye
sns.set_theme(style="whitegrid")

print("✅ All libraries are succesfuly imported!")
```

```python
df = pd.read_csv("../data/reviews.csv")

df.head()
```

```python
print('Dataset Shape:', df.shape)
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# Custom figure size define kar rahe hain
plt.figure(figsize=(8, 5))

# Seaborn countplot se graph generate kar rahe hain
sns.countplot(data=df, x='Rating', palette='viridis')

# Graph ke titles aur labels add karna
plt.title('Distribution of Product Ratings', fontsize=14, fontweight='bold')
plt.xlabel('Rating (Stars)', fontsize=12)
plt.ylabel('Count of Reviews', fontsize=12)

# Chart display karna
plt.show()
```

```python
df['Review_Length'] = df['Review_Text'].apply(len)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Rating', y='Review_Length', palette='viridis')

plt.titel('Review Length Distribution by Rating', fontsize=14, fontweight='bold')
plt.xlabel('Rating (Stars), fontsize=12')
plt.ylabel('Review Character Length', fontsize=12)
plt.show()
```
